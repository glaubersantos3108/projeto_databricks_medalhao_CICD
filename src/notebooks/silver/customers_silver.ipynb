```python
from pyspark.sql.functions import *
```

```python
%py
df_customers_silver = spark.sql("""
    SELECT
    a.customer_id
    ,a.first_name
    ,a.last_name
    ,LTRIM(RTRIM(a.phone)) AS phone
    ,a.email
    ,a.street
    ,a.city
    ,a.state
    ,a.zip_code
    FROM lugus_databricks_dev.bronze.customers a
    WHERE NULLIF(LTRIM(RTRIM(a.phone)), '') IS NOT NULL
    AND a.phone <> 'NULL'
""")
```

```python
%py
df_customers_silver.write.mode("overwrite").format("delta").options(
    overwriteSchema = "true"
).saveAsTable('lugus_databricks_dev.silver.customers_silver')
```
