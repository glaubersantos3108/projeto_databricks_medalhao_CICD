```python
catalog = dbutils.widgets.get("catalog")
```

```python
%py
df_sales_gold = spark.sql(f"""
    SELECT
     a.order_id
    ,a.customer_id
    ,a.order_status
    ,SUM(a.quantity) AS quantity
    ,a.order_date
    ,a.store_name
    ,a.state
    ,a.city
    ,a.required_date
    ,a.shipped_date
    ,a.fullname
    ,a.active
    ,a.email
    ,ROUND(SUM(a.total_amount),2) AS total_amount
    ,a.total_amount_discount
    ,a.discount
    FROM {catalog}.silver.orders a
    WHERE a.shipped_date IS NOT NULL
    GROUP BY
     a.order_id
    ,a.customer_id
    ,a.order_status
    ,a.order_date
    ,a.store_name
    ,a.state
    ,a.city
    ,a.required_date
    ,a.shipped_date
    ,a.fullname
    ,a.active
    ,a.email
    ,a.total_amount_discount
    ,a.discount
""")
```

```python
%py
df_sales_gold.write.mode("overwrite").format("delta").options(
    overwriteSchema = 'true'
).saveAsTable(f"{catalog}.gold.sales")
```
