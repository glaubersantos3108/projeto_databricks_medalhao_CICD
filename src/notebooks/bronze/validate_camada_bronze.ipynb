```python
%py
df_tabelas_silver = spark.sql("""
    SELECT 
    CONCAT(table_catalog, '.', table_schema, '.', table_name) AS table_name
    FROM lugus_databricks_dev.information_schema.tables
    WHERE table_schema = 'silver'
""")

for table_name in df_tabelas_silver.collect():
  spark.sql(f"TRUNCATE TABLE {table_name[0]}")
```
